# PLOTY LIBRARY/PACKAGE

### Plotly in Python is a data visualization library used to create interactive, high-quality, and web-based charts.

## Key Uses of Plotly in Python

### Interactive Visualizations

1. Unlike static plots (e.g., Matplotlib), Plotly charts allow zooming, panning, hovering for tooltips, and dynamic updates.
Wide Range of Chart Types
2. Supports line charts, bar charts, scatter plots, pie charts, histograms, heatmaps, 3D plots, maps, and more.
Web & Dashboard Integration
3. Works seamlessly with Dash (a Python framework) to build interactive dashboards without needing JavaScript.
Publication-Quality Graphics
4. Produces visually appealing, professional-grade plots suitable for reports and presentations.
Multiple Output Formats

5. Can display in Jupyter Notebooks, save as HTML, or embed in web applications.

# MATPLOTLIB LIB/PACKAGE

- Matplotlib is a popular data visualization library in Python used to create a wide variety of static, interactive, and animated plots.

### Main Uses of Matplotlib
1. Data Visualization
  - Helps represent data in a graphical format (line charts, bar charts, scatter plots, histograms, pie charts, etc.).
Makes it easier to understand trends, patterns, and relationships in data.
2. Customizable Plots
  - You can control every aspect of a plot: colors, labels, titles, legends, grid lines, and styles.
3. Supports Multiple Output Formats
  - Can save plots as PNG, JPG, SVG, PDF, etc.
4. Works in Jupyter Notebooks, Python scripts, and interactive environments.
5. Integration with Other Libraries
- Works well with NumPy, Pandas, and SciPy for scientific computing.
Often used alongside Seaborn for more advanced statistical visualizations.
