### 5.Understand the Dataset Using info()
```
df.info()

The info() method displays:
Column names
Number of non-null values
Data type of every column
Total memory usage

Example data types:
int64 – whole numbers
float64 – decimal numbers
object – text or mixed values
datetime64 – date and time values

In this dataset, Order Date and Ship Date may initially appear as object.
We will convert them into proper date columns later.
```

````python
store_details.info()
````

### 6.Check Data Types
dtype is a **instance**
```
df.dtypes

To display the column name and data type clearly:
for column in df.columns:
    print(column, ":", df[column].dtype)
    
Understanding data types is important because different operations require different data types.

For example:
We cannot perform date calculations properly on a text column.
We cannot calculate averages on a product name column.
We need numerical columns for mathematical analysis.
```

````python
store_details.dtypes
````

````python
for column in store_details.columns:
    print(column,":",store_details[column].dtype)
````
