## <font color="cyan">Handling the Missing Values</font>
```
Method:01
-using dropna()
-use it carefully because it remove the missing value permanently
eg; df["Sales"]=df["sales"].dropna(inplace=True)

Method:02
-using fillna().mean()
-use fillna() to fillnumerical value at missing place
eg; df["sales"]=df["sales"].fillna(df["sales"].mean())

Method 3: 
-Fill numerical missing values using the median
eg;df["Profit"] = df["Profit"].fillna(df["Profit"].median())

Method 4:
-Fill categorical missing values using the mode
eg;df["Region"] = df["Region"].fillna(df["Region"].mode()[0])
[0] because it return first indexed repeatative value that means mode Value

Method 5: 
-Fill missing values with a custom value
eg;df["Region"] = df["Region"].fillna("Unknown")

Choosing a method
Use the mean when values are normally distributed.
Use the median when outliers are present.
Use the mode for categorical values.
Remove rows only when missing records are very few and unimportant.
```

````python
stu_details.dtypes
````

````python
for col in stu_details.columns:
    stu_details[col]=stu_details[col].fillna(stu_details[col].mode()[0])
stu_details.head(50)
````
