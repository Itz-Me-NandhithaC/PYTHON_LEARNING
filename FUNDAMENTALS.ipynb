# Number Guessing using while loop (infinte)
### Conditions
1. 7 trials
2. the number is betweeN 1 to 100
3. if user entered invalid number (like -ve number) then program should says "number is invalid"
4. if user number > actual number then, program should says " Your number is MAXIMUM "
5. if user number < actual number then,porgram should says "Your number is MINIMUM "

```python
secret_number=45
print("Hey,Welcome,")
print("Guess my number the range between 1 to 100")
for i in range(0,7):
    print("Trial: ",i+1)
    user_input=int(input("Enter the Number: "))
    if user_input < 0 or user_input > 100:
        print("Invalid number,Please enter the Number between 1 to 100")
    elif user_input>secret_number:
        print("Your number is Higher")
    elif user_input<secret_number:
        print("Your number is Low")
    else:
        print("Congratualtions!!🎊")
        break
```

# Using the "random" library same condition applied on number Guessing Number

```python
import random
secret_number=random.randint(1,100)
print("Welcome,Guessing Number Game🧩")
print("📌Guess 1 to 100")

for i in range(10):
    print("Trial:",i+1,"Remaining trails",10-i)
    user_input=int(input("Enter the number:"))
    if user_input < 0 or user_input > 100:
        print("❌Invalid, Please enter number between 1 to 100")
    elif secret_number > user_input:
        print("Secret number is higher than you thinks")
    elif secret_number < user_input:
        print("Secret number is less than you thinks")

    else:
        print("Your Guessed number is Correct!!")
        break
if secret_number == user_input:
    print(f"Congratulation, You found {secret_number} at trial {i+1}" )
else:
    print("Sorry,You have reached limit")
```

# Function ( run program as a function )
1. define funtion
2. call the function
3. "Yes to CREATE IT"

```python
def number_guessing_game():
    import random
    secret_number=random.randint(1,100)
    print("Welcome,Guessing Number Game")
    print("Guess 1 to 100")
    
    for i in range(10):
        try:
            print("Trial:",i+1,"Remaining trails",10-i)
            user_input=int(input("Enter the number:"))
        
            if user_input < 0 or user_input > 100:
                print("Invalid, Please enter number between 1 to 100")
            elif secret_number > user_input:
                print("Secret number is higher than you thinks")
            elif secret_number < user_input:
                print("Secret number is less than you thinks")
        
            else:
                print("Your Guessed number is Correct!!")
                break
        except ValueError:
            print("Hey,,,, please enter Number from 1 to 100")
    if secret_number == user_input:
        print(f"Congratulation, You found {secret_number} at trial {i+1}" )
    else:
        print("Sorry,You have reached limit")
```

## Now CALL IT 👇👇

```python
number_guessing_game()
```
